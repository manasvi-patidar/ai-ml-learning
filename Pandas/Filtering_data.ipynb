```python
import pandas as pd
```

```python
df = pd.read_csv("GlobalAirQuality.csv")
```

```python
#Filtering of Data

aqi_data = df[ (df["aqi"] > 100) & (df["temperature"] > 30) ][["city", "aqi"]]
```

```python
#Filtering data using Query()

df.query("aqi > 200")
```

```python
df.query("aqi > 200 & temperature < 30")[["city", "aqi"]]
```

```python
#variables in query()
aqi_val = 200
df.query("aqi > @aqi_val & temperature < 30")[["city", "aqi"]]
```
