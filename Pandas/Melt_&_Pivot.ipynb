```python
import pandas as pd
```

```python
# Melt & Pivot

df = pd.DataFrame({
    "country" : ["USA", "USA", "India", "India"],
    "year" : [2024, 2025, 2024, 2025],
    "sales" : [100, 120, 90, 110],
    "profit" : [20, 25, 18, 22],
    "tax" : [10, 20, 30, 40]
})
```

```python
df
```

```python
# Melt

melted_df = df.melt(
    id_vars=["country", "year"], #keep
    value_vars=["sales", "profit", "tax"], #unpivot
    var_name="metrics",
    value_name="value"
)
```

```python
melted_df
```

```python
# Pivot

pivoted_df = melted_df.pivot(
    index=["country", "year"],
    columns="metrics",
    values="value"
)
```
