```python
import pandas as pd
```

```python
# Concatenation

df1 = pd.DataFrame({
    "id" : [1, 2, 3],
    "name" : ["Adam", "Eve", "Bob"]
})

df2 = pd.DataFrame({
    "id": [4, 5, 6],
    "name" : ["Charlie", "Fester", "Gigi"]
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2]) # row
```

```python
pd.concat([df1, df2], ignore_index=True) #override original labels
```

```python
pd.concat([df1, df2], axis=1) # col
```
