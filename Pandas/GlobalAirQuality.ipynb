```python
import pandas as pd
```

```python
df = pd.read_csv("GlobalAirQuality.csv")
df.head()
```

```python
#selecting data

#Columns
df["city"]
df[["city", "aqi"]]
```

```python
#Rows

#Label-based access
df.loc[0] #first row data
```

```python
df.loc[0:2] #start idx : end idx (ending index i.e. 2 is inclusive in case of loc)
```

```python
#Position-based access

df.iloc[2]
df.iloc[0:2] # 2 is exclusive in case of iloc
```

```python
#indivial cells (rows, columns)

df.loc[0, "aqi"]
```

```python
df.loc[0, ["city", "aqi"]]
```

```python
df.loc[0:2, ["city", "aqi"]]
```

```python
#iloc cannot access str directly, its index has to be passed

df.iloc[0:3, 2] #idx of city -> 2
```

```python
df.loc[0:2, ["country", "city", "latitude", "longitude"]] # str can be accessed
```

```python
df.iloc[0:3, 1:5] #str can't be accessed. idx has to be passed
```

```python
# Access single scalar value (cell)

df.at[0, "city"]
```

```python
df.iat[0, 2]
```
