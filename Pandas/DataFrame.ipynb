```python
import pandas as pd
```

```python
#Creating DataFrame

#dictionary of lists
info = {
    "Name": ["Adam", "Eve", "Bob"],
    "Age": [23, 24, 25],
    "CGPA": [9.5, 8.6, 7.2]
}

df = pd.DataFrame(info)

print(df, type(df))

print(df.index)
print(df.columns)
```

```python
#lists of lists
df = pd.DataFrame([["Adam", 23], ["Eve", 24], ["Bob", 25]], columns=["Name", "Age"])

print(df)
```

```python
#np arrays

import numpy as np

np_arr = np.array([[1, 2, 3], [4, 5, 6]])

df = pd.DataFrame(np_arr, columns=["A", "B", "C"])

print(df)
```
