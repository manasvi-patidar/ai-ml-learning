```python
import pandas as pd
```

```python
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

#index
print(s[0])
print(s[2])
```

```python
s = pd.Series(
    [21, 22, 23, 24, 25],
    index=["Adam", "Eve", "Casey", "Bob", "David"]
)

print("Complete Series:")
print(s)

print("\nLabel-based access:")
print(s.loc["Casey"])

print("\nPosition-based access:")
print(s.iloc[0])
```

```python
#Vectorized Operations in Series
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([10, 20, 30, 40, 50])

print(s1 + s2)

#Mutable
s1[0] = 100
changed_s1 = s1.drop(0)

print(s1)
print(changed_s1)
```
