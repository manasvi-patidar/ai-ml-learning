```python
import numpy as np
```

```python
#Mathematical functions

#aggregation fnx
arr = np.array([1, 2, 3, 4, 5])

print(np.sum(arr))
print(np.prod(arr))

print(np.min(arr))
print(np.max(arr))

print(np.argmin(arr)) #returns idx of min val
print(np.argmax(arr)) #returns idx of max val

print(np.mean(arr))
print(np.median(arr))

print(np.std(arr))
print(np.var(arr))
```

```python
#power fnx
arr = np.array([1, 2, 3, 4, 5])

print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr, 3))
```

```python
#log fnx
arr = np.array([1, 2, 3, 4, 5])

print(np.log(arr))
print(np.log2(arr))
print(np.log10(arr))
```

```python
#exponential fnx
arr = np.array([1, 2, 3, 4, 5])

print(np.exp(arr))
```

```python
#rounding fnx
print(np.round(3.14))
print(np.ceil(3.14))
print(np.floor(3.14))
print(np.trunc(3.14))
```

```python
#extra fnx
arr = np.array([1, 2, 8, 3, 4, 5, 3, 2])
arr2 = np.array([1, 2, -8, -3, -7])

print(np.unique(arr))
print(np.sort(arr))
print(np.abs(arr2))
```
