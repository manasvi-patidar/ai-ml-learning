```python
import numpy as np
```

```python
#Data Types

#int
arr = np.array([1, 2, 3, 4, 5])
print(arr, arr.dtype)
```

```python
#float
arr = np.array([1, 2, 3.14, 4, 5])
print(arr, arr.dtype)
```

```python
#string
arr = np.array([1, 2, 3, 4, 5, "manasviii"])
print(arr, arr.dtype)
```

```python
#complex no.s
arr = np.array([3 + 5j])
print(arr, arr.dtype)

arr1 = np.array([3 + 5j])
arr2 = np.array([2 + 3j])
print(arr1 + arr2)
```
