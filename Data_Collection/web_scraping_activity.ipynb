```python
import requests
from bs4 import BeautifulSoup
```

```python
#Fetch data from all pages

page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore...")
        break

    with open(f"scraped_data/quotes{page_count}.html", "w" ,encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded data from page {page_count}")

    page_count += 1
```

```python
# Extract life quotes from ALL downloaded pages
life_quotes = []

for page in range(1, page_count):
    with open(
        f"scraped_data/quotes{page}.html",
        "r",
        encoding="utf-8"
    ) as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")

    for q in all_quotes:
        all_tags = [
            tag.get_text(strip=True)
            for tag in q.select(".tags .tag")
        ]

        if "life" in all_tags:
            text = q.select_one("span.text").get_text(strip=True)
            author = q.select_one("small.author").get_text(strip=True)

            life_quotes.append([text, author])
```

```python
import pandas as pd

df = pd.DataFrame(life_quotes, columns=["Quote", "Author"])

df.to_csv("cleaned_data/quotes.csv", index=False, encoding="utf-8")

print("Life quotes saved successfully!")
```
