```python
import matplotlib.pyplot as plt
```

```python
# Monthly Business Expenses
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amount = [500, 150, 120, 100, 50]

plt.pie(amount, labels=expenses)
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amount = [500, 150, 120, 100, 50]

plt.style.use("seaborn-v0_8-whitegrid")
plt.pie(amount, labels=expenses, autopct="%1.1f%%")
plt.show()
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amount = [500, 150, 120, 100, 50]

colors = ["#5B9A91", "#7895B2", "#9CAF88", "#D8B887", "#CC8B78"]

plt.style.use("seaborn-v0_8-whitegrid")

plt.pie(amount, 
        labels=expenses,
        colors=colors, 
        autopct="%1.1f%%",
        wedgeprops = {"edgecolor": "white", "linewidth": 1.5}
       )
plt.show()
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amount = [500, 150, 120, 100, 50]

colors = ["#5B9A91", "#7895B2", "#9CAF88", "#D8B887", "#CC8B78"]

plt.style.use("seaborn-v0_8-whitegrid")

plt.pie(amount, 
        labels=expenses,
        colors=colors, 
        autopct="%1.1f%%",
        wedgeprops = {"edgecolor": "black", "linewidth": 1.5, "fill": False}
       )
plt.show()
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amount = [500, 150, 120, 100, 50]

colors = ["#5B9A91", "#7895B2", "#9CAF88", "#D8B887", "#CC8B78"]

explode = [0, 0, 0, 0.1, 0]

plt.style.use("seaborn-v0_8-whitegrid")

plt.pie(amount, 
        labels=expenses,
        colors=colors, 
        autopct="%1.1f%%",
        wedgeprops = {"edgecolor": "white", "linewidth": 1.5},
        explode=explode
       )
plt.show()
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amount = [500, 150, 120, 100, 50]

colors = ["#5B9A91", "#7895B2", "#9CAF88", "#D8B887", "#CC8B78"]

explode = [0, 0, 0, 0.1, 0]

plt.style.use("seaborn-v0_8-whitegrid")

plt.pie(amount, 
        labels=expenses,
        colors=colors, 
        autopct="%1.1f%%",
        wedgeprops = {"edgecolor": "white", "linewidth": 1.5},
        explode=explode,
        startangle=90,
        shadow=True
       )
plt.show()
```
