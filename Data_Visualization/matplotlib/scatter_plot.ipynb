```python
import matplotlib.pyplot as plt
```

```python
# Customizations

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 135 else "red" for x in bp]

plt.scatter(age, bp, color=colors, s=bp, marker="o", alpha=0.5)
plt.title("Age vs. Blood Pressure")
plt.xlabel("Age (years)")
plt.ylabel("Blood Pressure (mmHg)")

plt.grid()
```

```python
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.scatter(age, bp, s=bp, cmap="OrRd", c=bp)
plt.title("Age vs. Blood Pressure")
plt.xlabel("Age (years)")
plt.ylabel("Blood Pressure (mmHg)")

plt.grid()
plt.colorbar(label="BP")
```

```python
# Annotations
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.figure(figsize=(10, 6))

plt.scatter(age, bp, s=bp, cmap="OrRd", c=bp)
plt.title("Age vs. Blood Pressure")
plt.xlabel("Age (years)")
plt.ylabel("Blood Pressure (mmHg)")

plt.grid(alpha=0.3)
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i]+1, bp[i]), fontsize=9)

# Add padding around the data
plt.xlim(min(age) - 5, max(age) + 10)
plt.ylim(min(bp) - 10, max(bp) + 10)
```

```python
# Multiple Data: Summer vs Winter

cities = ["City A", "City B", "City C", "City D", "City E"]

winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.figure(figsize=(9, 5))

plt.scatter(
    winter_temp, winter_humidity,
    color="#2A9D8F",
    s=100,
    edgecolors="white",
    linewidths=1,
    label="Winter",
    alpha=0.9
)

plt.scatter(
    summer_temp, summer_humidity,
    color="#E76F51",
    s=100,
    edgecolors="white",
    linewidths=1,
    label="Summer",
    alpha=0.9
)

plt.title("Summer vs. Winter: Temperature and Humidity",
          fontsize=14, fontweight="bold")
plt.xlabel("Temperature (°C)", fontsize=11)
plt.ylabel("Humidity (%)", fontsize=11)

plt.legend(frameon=True)
plt.grid(linestyle="--", alpha=0.3)

plt.xlim(-5, 40)
plt.ylim(35, 95)

plt.tight_layout()
plt.show()
```
