```python
import matplotlib.pyplot as plt
```

```python
Oscar_Movies = [
        "The Dark Knight",
        "The Hurt Locker",
        "The King's Speech",
        "The Artist",
        "Argo"
    ]

Year = [2008, 2009, 2010, 2011, 2012]
Oscar_Revenue = [1005, 170, 427, 133, 232]

plt.bar(Year, Oscar_Revenue, color="#2A9D8F")

plt.title("Oscar-Winning Movies: Revenue Over Time")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
```

```python
Year = [2008, 2009, 2010, 2011, 2012]
Oscar_Revenue = [1005, 170, 427, 133, 232]

plt.bar(Year, Oscar_Revenue, color="#2A9D8F")

plt.title("Oscar-Winning Movies: Revenue Over Time")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

for i in range(len(Year)):
    plt.text(Year[i], Oscar_Revenue[i]+20, str(Oscar_Revenue[i]), ha="center")
```

```python
Year = [2008, 2009, 2010, 2011, 2012]
Oscar_Revenue = [1005, 170, 427, 133, 232]

plt.bar(Year, Oscar_Revenue, color="#2A9D8F")

plt.title("Oscar-Winning Movies: Revenue Over Time")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

for i in range(len(Year)):
    plt.text(Year[i], Oscar_Revenue[i]+20, str(Oscar_Revenue[i]), ha="center")

plt.ylim(0, max(Oscar_Revenue)+100)
```

```python
import numpy as np

Oscar_Revenue = [1005, 170, 427, 133, 232]
Non_Oscar_Revenue = [378, 2788, 829, 185, 275]

Year = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(Year))
width = 0.4

plt.bar(x-width/2, Oscar_Revenue, width, label="Oscar Movies") # shifting to left
plt.bar(x+width/2, Non_Oscar_Revenue, width, label="Non-Oscar Movies") # shifting to right

plt.title("Oscar vs Non-Oscar Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.legend()
```

```python
Oscar_Revenue = [1005, 170, 427, 133, 232]
Non_Oscar_Revenue = [378, 2788, 829, 185, 275]

Year = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(Year))
width = 0.4

plt.bar(x-width/2, Oscar_Revenue, width, label="Oscar Movies")
plt.bar(x+width/2, Non_Oscar_Revenue, width, label="Non-Oscar Movies")

plt.title("Oscar vs Non-Oscar Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.legend()

plt.xticks(x, Year)
```

```python
# Horizontal Bar Chart

Oscar_Movies = [
        "The Dark Knight",
        "The Hurt Locker",
        "The King's Speech",
        "The Artist",
        "Argo"
    ]

Oscar_Revenue = [1005, 170, 427, 133, 232]

plt.barh(Oscar_Movies, Oscar_Revenue)

plt.title("Oscar-Winning Movies: Revenue Over Time")
plt.xlabel("Revenue (in $M)")
plt.ylabel("Year")

plt.tight_layout()
plt.show()
```
