```python
import matplotlib.pyplot as plt
```

```python
X = [1, 2, 3, 4]
Y = [5, 6, 7, 8]

plt.plot(X, Y) # line plot
plt.show()
```

```python
Oscar_Movies = [
        "The Dark Knight",
        "The Hurt Locker",
        "The King's Speech",
        "The Artist",
        "Argo"
    ]

Year = [2008, 2009, 2010, 2011, 2012]
Oscar_Revenue = [1005, 170, 427, 133, 232]

plt.plot(Year, Oscar_Revenue)

plt.title("Oscar-Winning Movies: Revenue Over Time")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
```

```python
Non_Oscar_Movies = [
        "Men in Black",
        "Avatar",
        "Inception",
        "Finding Nemo",
        "Barbie"
    ]

Non_Oscar_Revenue = [378, 2788, 829, 185, 275]

plt.plot(Year, Oscar_Revenue, label="Oscar Movies")
plt.plot(Year, Non_Oscar_Revenue, label="Non-Oscar Movies")

plt.title("Oscar vs Non-Oscar Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.legend()

plt.savefig("final_plot.png")
```

```python
# Format Strings

plt.plot(Year, Oscar_Revenue, "o--r", label="Oscar Movies")
plt.plot(Year, Non_Oscar_Revenue, "<-y", label="Non-Oscar Movies")
```

```python
plt.plot(Year, Oscar_Revenue, color="red", marker="o", linestyle="-.", linewidth=2, label="Oscar Movies")
plt.plot(Year, Non_Oscar_Revenue, label="Non-Oscar Movies")
```
